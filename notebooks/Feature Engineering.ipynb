```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv(r"C:\Users\Lenovo\Desktop\intern project 2\Cleaned\preprocessed_policy_dataset.csv")

df.head(100)
```

```python
df.shape
```

```python
df["impact_label"].value_counts()
```

```python
df.drop_duplicates(subset=["processed_text"])
```

```python
X_text = df["processed_text"]
```

```python
vectorizer = TfidfVectorizer(max_features=5000)

X = vectorizer.fit_transform(X_text)

print("Feature matrix shape:", X.shape)
```

```python
feature_names = vectorizer.get_feature_names_out()

X_df = pd.DataFrame(X.toarray(), columns=feature_names)

X_df.head()
```

```python
final_dataset = pd.concat(
    [
        X_df,
        df[["policy_area", "public_sentiment", "impact_label"]].reset_index(drop=True)
    ],
    axis=1
)

final_dataset.head()
```

```python
final_dataset.to_csv(r"C:\Users\Lenovo\Desktop\intern project 2\Processed Data\feature_engineered_dataset.csv", index=False)

print("Feature engineered dataset saved successfully")
```
